```python
plt.figure(figsize = (6,4))
sns.countplot(x = 'target', data = df_train)
plt.show()

# 1 - клиент просрочил платеж
```

```python
corr = df_train.select_dtypes(include=['float64', 'int64']).corr()['target'].sort_values()
print("Топ-10 признаков с положительной корреляцией:\n", corr.tail(11))
print("Топ-10 признаков с отрицательной корреляцией:\n", corr.head(10))
```

```python
plt.figure(figsize=(10, 6))
corr[abs(corr) > 0.04].sort_values().plot.barh()
plt.title('Корреляция с TARGET')
plt.show()
```
