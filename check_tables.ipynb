```python
credits_per_client = df_bruean['sk_id_curr'].value_counts()
plt.figure(figsize=(10, 5))
sns.histplot(credits_per_client, bins=50, kde=True)
plt.title('Сколько кредитов у одного клиента')
plt.xlabel('Количество кредитов')
plt.ylabel('Количество клиентов')
plt.show()

credits_per_client.describe().T
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df_bruean, x='credit_active', order=df_bruean['credit_active'].value_counts().index)
plt.title('Статус кредитов в бюро')
plt.xlabel('Статус')
plt.ylabel('Количество')
plt.show()

df_bruean['credit_active'].value_counts()
```

```python
df_bruean['has_overdue'] = (df_bruean['credit_day_overdue'] > 0)
plt.figure(figsize=(6, 4))
sns.countplot(data=df_bruean, x='has_overdue')
plt.title('Есть просрочка?')
plt.xlabel('Просрочка (0 = нет, 1 = да)')
plt.show()
```
