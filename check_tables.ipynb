```python
payments_per_credit = df_i_p['sk_id_prev'].value_counts()

plt.figure(figsize=(12,5))
sns.histplot(payments_per_credit, bins=50)
plt.title('Количество платежей на кредит')
plt.xlabel('Платежей')
plt.xlim(0, 60)
plt.show()
print(payments_per_credit.describe())
```

```python
df_i_p['dpd'] = df_i_p['days_instalment'] - df_i_p['days_entry_payment']
df_i_p['dpd'] = df_i_p['dpd']

dpd_pos = df_i_p[df_i_p['dpd'] > 0]['dpd']
plt.figure(figsize=(12,5))
sns.histplot(dpd_pos, bins=50, log_scale=(False, True))
plt.title('Просрочки')
plt.xlabel('DPD')
plt.show()

print("Месяцев с просрочкой:", (df_i_p['dpd'] > 0).sum())
print("Макспросрочка:", df_i_p['dpd'].max())
```

```python
under = df_i_p[df_i_p['underpayment'] > 0]['underpayment']
plt.figure(figsize=(12,5))
sns.histplot(under, bins=50, log_scale=(False, True))
plt.title('Недоплаты')
plt.xlabel('Сумма недоплаты')
plt.show()

print('Сумма недопла:', under.sum())
```
